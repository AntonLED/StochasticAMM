Lets consider portfolio with $\alpha$ in numeraire ($x$) and $(1 - \alpha)$ in risky ($y$).

$p$ -- price of risky asset in numeraire asset.

---

### Invariant for $\alpha$-portfolio

Portfolio s.t. conditions:

$$
\begin{cases}
  V = x + p y, \\
  \alpha V = x \cdot 1, \\
  (1 - \alpha) V = y \cdot p.
\end{cases}
$$

So we have a differential equation:

$$
\begin{aligned}
  \frac{dy}{dx} &= -\frac{\alpha}{1 - \alpha} \cdot \frac{y}{x}, \\
  \log{y} &= -\frac{\alpha}{1 - \alpha} \log{x} + C.
\end{aligned}
$$

And finally gets an invariant (something familiar XD):

$$
y \cdot x^{\frac{\alpha}{1 - \alpha}} \equiv \left/ \text{according to dimensions}: \; D^{1 + \frac{\alpha}{1 - \alpha}} \right/ = D^{\frac{1}{1 - \alpha}},
$$

$$
\boxed{y^{1 - \alpha} x^{\alpha} = D}.
$$

It is the **_Balancer invariant_**.

---

### $\frac{dp}{p} \Longleftrightarrow d\log{p}$

For log-prices we assume the brownian motion:

$$
d \log{p} = \mu \, dt + \sigma \, dW_t.
$$

So for $p$ we should substitute $f(x, t) = \exp{x}$ in Itô formula:

$$
df(x, t) = \left(\partial_t f + \mu \, \partial_x f + \frac{\sigma^2}{2} \partial_x^2 f\right) dt + \sigma \, \partial_x f \, dW_t.
$$

So for $p$ evolution we have:

$$
\begin{aligned}
  dp &= \left( \mu \exp{(\log{p})} + \frac{\sigma^2}{2} \exp{(\log{p})} \right) dt + \sigma \exp{(\log{p})} \, dW_t
  = p \left(\mu + \frac{\sigma^2}{2}\right) dt + p \sigma \, dW_t \\
  &\Longleftrightarrow \frac{dp}{p} = \left( \mu + \frac{\sigma^2}{2} \right) dt + \sigma \, dW_t.
\end{aligned}
$$

We have an **_additional drift_**.

### Invariant evolution for the $\alpha$-portfolio

Lets assume finite price jump $\Delta p$ has occurred. For $\alpha$-portfolio we have equations:

$$
\begin{cases}
  y_1^{1 - \alpha} x_1^{\alpha} = D_1, \\
  y_2^{1 - \alpha} x_2^{\alpha} = D_2, \\
  x_1 = \frac{\alpha}{1 - \alpha} p y_1, \\
  x_2 = \frac{\alpha}{1 - \alpha} (p + \Delta p) y_2, \\
  (x_2 - x_1) = -(p + \Delta p)(y_2 - y_1).
\end{cases}
$$

So for $D_2$ and $D_1$ relation we have:

$$
\begin{aligned}
  \boxed{\frac{D_2}{D_1}}
  &= \left(\frac{p + \Delta p}{p}\right)^{\alpha} \cdot \frac{y_2}{y_1}
  = \left(\frac{p + \Delta p}{p}\right)^{\alpha} \cdot \frac{\frac{1}{1 - \alpha} p + \Delta p}{\frac{1}{1 - \alpha} (p + \Delta p)} \\
  &= \left/ \; \gamma \equiv \Delta p / p \; \right/
  = \boxed{\frac{1 + (1 - \alpha) \gamma}{(1 + \gamma)^{1 - \alpha}}}.
\end{aligned}
$$

According to the Taylor formula $(1 + x)^{\alpha} = \sum_{n = 0}^{\infty} \binom{\alpha}{n} x^n = 1 + \alpha x + \frac{\alpha (\alpha - 1)}{2!} x^2 + \mathcal{O}(x^3)$ (for $|x| < 1$ ofc) we have asymptotic:

$$
\boxed{\frac{D_2}{D_1} = 1 + \frac{\alpha (1 - \alpha)}{2} \gamma^2 + \mathcal{O}(\gamma^3)}.
$$

Now lets write this evolution of $D$ not only in terms of linear delta but also in log-delta:

$$
\begin{aligned}
  \gamma_\text{lin} &= \frac{\Delta p}{p} \quad (\equiv \gamma), \\
  \gamma_\text{log} &= \log{\frac{p + \Delta p}{p}} \quad (\equiv \Delta S).
\end{aligned}
$$

$$
\left[
\begin{aligned}
  &\text{linear}: &
  \frac{D_2}{D_1}
  &= \frac{1 + (1 - \alpha) \gamma_\text{lin}}{(1 + \gamma_\text{lin})^{1 - \alpha}}
  = 1 + \frac{\alpha (1 - \alpha)}{2} \gamma_\text{lin}^2 + \mathcal{O}(\gamma_\text{lin}^3), \\
  &\text{logarithmic}: &
  \frac{D_2}{D_1}
  &= (1 - \alpha) \exp{(\alpha \gamma_\text{log})} + \alpha \exp{(-(1 - \alpha) \gamma_\text{log})}
  = e^{\gamma_\text{log} \left( \alpha - \frac{1}{2} \right)} \left[ \cosh{\frac{\gamma_\text{log}}{2}} - 2 \left( \alpha - \frac{1}{2} \right) \sinh{\frac{\gamma_\text{log}}{2}} \right] \\
  & & &= 1 + \frac{\alpha (1 - \alpha)}{2} \gamma_\text{log}^2 + \mathcal{O}(\gamma_\text{log}^3).
\end{aligned}
\right.
$$

### $\log{\frac{D}{D_0}}$ deviation with drift included ($\mu \neq 0$)

Now we return to $50/50$-portfolio (check the [lecture](./seminar-2.pdf)).

According to the [lecture](./seminar-2.pdf) (and also to the expressions above) one can write:

$$
\log{\frac{D}{D_0}} = \frac{1}{8} \alpha^2 + \mathcal{O}(\alpha^3), \quad \alpha = \Delta S.
$$

Now assume $N$ little steps in time $T$ so each step has $\Delta t = \frac{T}{N}$ duration.

Also make for $S$ the traditional assumptions:

$$
\Delta S_i = \mu \Delta t + \sigma \sqrt{\Delta t} \, \xi_i, \quad \xi_i \sim \mathcal{N}(0, 1)
\;\Longrightarrow\;
\Delta S_i \equiv \alpha_i \sim \mathcal{N}(\mu \Delta t, \sigma^2 \Delta t).
$$

We now can write:

$$
\log{\frac{D_{i + 1}}{D_i}} = \frac{1}{8} \alpha_i^2 + \mathcal{O}(\alpha_i^3)
\;\Longrightarrow\;
\log{\frac{D_N}{D_0}} = \sum_{i = 0}^{N - 1} \frac{1}{8} \alpha_i^2.
$$

For now let's do some probability stuff:

$$
\mathbb{E} \alpha_i^2 = \mathbb{D} \alpha_i + (\mathbb{E} \alpha_i)^2 = \sigma^2 \Delta t + \mu^2 \Delta t^2.
$$

Finally:

$$
\mathbb{E} \log{\frac{D_N}{D_0}} \equiv \mathbb{E} \log{\frac{D}{D_0}}
= \frac{N}{8} \mathbb{E} \alpha_i^2
= \frac{\sigma^2}{8} T + \frac{\mu^2}{8} T \Delta t
\xrightarrow[\Delta t \to 0]{} \frac{\sigma^2}{8} T.
$$

So the **_drift term_** $\mu$ **_does not affect on expected invariant growth_**.

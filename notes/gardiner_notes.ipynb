- generating function (производящая функция) $G(s, t) = \sum s^n P(n, t)$ для вероятности $P(n, t)$. С помощью этого чел решал master equation $\frac{\partial P(n, t)}{\partial t} = \lambda \left[ P(n - 1,t) P(n, t) \right] \iff \frac{\partial G(s, t)}{\partial t} = \lambda (s - 1) G(s, t)$.
